```python
import pyspark.sql.functions as F
from common_utils.logging import get_logger

logger = get_logger('ds2b_sqlserver')
```

```python
#read data from raw data volume
#add audit columns and metadata = last_updated_ts, file_path
#define target
#write data in delta tables

"""
basic - python print

advanced - python logger
"""
```

```python
raw_path = "/Volumes/retaildataplatform/bronze/raw_data/sqlserver_customers/load_date=2026-09-07/"
logger.info(f"Raw Path is defined at {raw_path}")

logger.info('Reading data')
df = spark.read\
        .format("csv")\
        .option("header","true")\
        .option("inferschema", "true")\
        .load(raw_path)

logger.info("sample data succesfully read %s", df.show())
```

```python

logger.info('adding audit columns')
df = df.withColumn("last_updated_ts", F.current_timestamp() )\
    .withColumn("file_path", F.col("_metadata.file_path"))

target_path = "retaildataplatform.bronze.sqlserver_customers"

logger.info(f"Target path = {target_path}")

logger.warning('Writing data')
df.write.format("delta")\
    .mode("overwrite")\
    .saveAsTable(target_path)
logger.info('Data Written succesfully at %s', target_path)
```
