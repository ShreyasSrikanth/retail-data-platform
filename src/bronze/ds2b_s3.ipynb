```python
raw_path = "/Volumes/retaildataplatform/bronze/raw_data/s3-bucket-ingestion/load_date=2026-09-11/"

df = spark.read.format("parquet").load(raw_path)
df.display()
```

```python
import pyspark.sql.functions as F

df = df.withColumn("last_updated_timestamp", F.current_timestamp() ) \
    .withColumn("file_path", F.col("_metadata.file_path"))
```

```python
target_path = "retaildataplatform.bronze.s3_bucket_ingestion"

df.write\
    .format("delta")\
    .mode("overwrite")\
    .saveAsTable(target_path)
```
