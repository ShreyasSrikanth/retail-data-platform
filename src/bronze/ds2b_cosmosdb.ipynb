```python
import logging

logger = logging.getLogger('ds2b_cosmosdb')

logger.setLevel(logging.INFO)

if not logger.handlers:
    handler = logging.StreamHandler()
    handler.setFormatter(logging.Formatter('%(asctime)s %(levelname)s %(message)s'))
    logger.addHandler(handler)

logger.info('')
logger.debug('debug solved')
logger.error('error occured')
```

```python
logger.info('Defining raw path')
raw_path = "/Volumes/retaildataplatform/bronze/raw_data/cosmosdb_sales_orders/test/load_date=2026-09-11/"
logger.info(f'raw path = {raw_path}')
```

```python
logger.info('reading data')
df = spark.read.format("json").load(raw_path)
logger.info(f'data read succesfully {df.show()}')
```

```python
import pyspark.sql.functions as F

df = df.withColumn("last_updated_ts", F.current_timestamp() )\
    .withColumn("file_path", F.col("_metadata.file_path"))


target_path = "retaildataplatform.bronze.cosmosdb_sales_orders"
logger.info('target path is %s', target_path)

logger.warning('data writing')
df.write.format("delta")\
    .mode("overwrite")\
    .saveAsTable(target_path)
logger.info('data successfully written in %s', target_path)
```
