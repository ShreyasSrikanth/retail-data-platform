```sql
Create or replace view customers as
select 
    lower(`customer_name`) AS customer_name,
    trim(split_part(lower(`customer_name`),',',2)) AS first_name,
    trim(split_part(lower(`customer_name`),',',1)) AS last_name,
    upper(`city`) AS city,
    regexp_replace(`postcode`,"\.0$","") AS postcode,
    from_unixtime(try_cast(`valid_from` as bigint)) as valid_from,
    from_unixtime(try_cast(`valid_to` as bigint)) as valid_to,
    "USA" AS country,
    district,
    lat,
    lon,
    number,
    loyalty_segment,
    last_updated_timestamp,
    state,
    street,
    region,
    tax_code,
    tax_id,
    units_purchased,
    unit
from retaildataplatform.bronze.sqlserver_customers;

select * from customers;

```
